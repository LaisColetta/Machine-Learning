# Diarization Tools Comparison
This notebook compares three speaker diarization tools:

**Pyannote** is an open-source Python library for speaker diarization, offering pre-trained models that segment audio based on speaker activity. It is highly accurate but requires some setup, including a Hugging Face token. Pyannote at [Pyannote Documentation](https://huggingface.co/pyannote/speaker-diarization).

**SpeechBrain** is a flexible, open-source toolkit for speech processing tasks, including speaker diarization. It provides pre-trained models, is easy to use, and is lightweight, making it suitable for both beginners and experienced users. [SpeechBrain Documentation](https://speechbrain.github.io/).

**Deepgram** is a commercial, API-based service that offers advanced speech recognition and diarization capabilities. It provides fast results and is highly accurate, though it requires an API key for access. [Deepgram API](https://www.deepgram.com/).


We will analyze the debate audio between Donald Trump and Kamala Harris in the 2024 elections. The main goal of this exercise is to compare these tree popular tools performance ([Source](https://www.assemblyai.com/blog/top-speaker-diarization-libraries-and-apis/)).


```python
#Load the mp3 file
audio_file = "/Users/laiscoletta/Desktop/DA_Final_Semester/Machine-Learning/TrumpHarrisDebate.mp3"
```

## Test 1: Pyannote

```python
from pyannote.audio import Pipeline

# Load pre-trained Pyannote diarisation model
pipeline = Pipeline.from_pretrained("pyannote/speaker-diarization", 
                                    use_auth_token="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")

# Perform diarisation
diarization = pipeline(audio_file)

# Map speakers to segments
pyannote_segments = []
for segment, _, speaker in diarization.itertracks(yield_label=True):
    pyannote_segments.append({
        "speaker": speaker,
        "start": segment.start,
        "end": segment.end
    })

print("Pyannote Diarisation Results:")
for seg in pyannote_segments:
    print(f"{seg['speaker']} from {seg['start']:.2f}s to {seg['end']:.2f}s")
```

## Test 2: SpeechBrain 

```python
# Import necessary modules from SpeechBrain
from speechbrain.pretrained import SpeakerDiarization
import torchaudio


# Load the pre-trained diarization model from SpeechBrain
diarization_model = SpeakerDiarization.from_hparams(source="speechbrain/spkdiarization-dihard", savedir="tmp_dir")

# Apply diarization to the audio file
diarization_result = diarization_model.diarize_file(audio_file)

# Print diarization result (output speakers and their time frames)
for segment, _, speaker in diarization_result:
    print(f"Speaker {speaker}: {segment.start:.2f} - {segment.end:.2f}")
```

## Test 3: Deepgram

```python
# Deepgram API Speaker Diarization
import requests

# Set up Deepgram API credentials
api_url = "https://api.deepgram.com/v1/listen"
api_key = "your_deepgram_api_key"

# Send audio file for processing
headers = {"Authorization": f"Token {api_key}"}
with open(audio_file, "rb") as f:
    response = requests.post(api_url, headers=headers, files={"file": f}, params={"diarize": "true"})

# Parse and display diarization results
diarization_data = response.json()

print("\nDeepgram Diarization Results:")
for transcript in diarization_data["results"]["channels"][0]["alternatives"]:
    for word in transcript["words"]:
        speaker = word.get("speaker", "Unknown")
        print(f"[{speaker}] {word['word']} ({word['start']:.2f}s - {word['end']:.2f}s)")
```
